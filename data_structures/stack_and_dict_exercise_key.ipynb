# Exercise: Neural Network Step Calculator

Deep learning frameworks such as `PyTorch` represent computations as graphs of operations and execute these operations according to their dependencies.

In this exercise, we build a simplified version of this idea using two basic data structures: a **dictionary** for storing values and a **stack** for evaluating operations.


## Objective

Implement a simple forward pass evaluator for a single neuron with a ReLU activation:

$$
y = \mathrm{ReLU}(w_1x_1 + w_2x_2 + b)
$$

The program should use:

* a **dictionary** called `memory` to store inputs, weights, and the bias
* a **stack** called `execution_stack` to evaluate the computation

## Task

Implement:

```python
evaluate_neuron(instructions, memory)
```

where:

* `memory` is a dictionary, mapping variable names to numerical values
* `instructions` describes the computation using `PUSH` and `OP` instructions

The following operations should be supported:

* `*` for multiplication
* `+` for addition
* `RELU` for $\max(0,x)$

The calculation must be performed using the stack. Do not use Python's built-in expression evaluation functionality.

You may assume that all inputs and instructions are valid. No edge cases are expected in this exercise.

## Example

INPUT:

````python
memory = {"x1": 2.0, "w1": 1.5, "x2": 4.0, "w2": -1.0, "b": 0.5}

instructions = [
    ("PUSH", "x1"),
    ("PUSH", "w1"),
    ("OP", "*"),
    ("PUSH", "x2"),
    ("PUSH", "w2"),
    ("OP", "*"),
    ("OP", "+"),
    ("PUSH", "b"),
    ("OP", "+"),
    ("OP", "RELU"),
]
````

RUN:
```python
evaluate_neuron(instructions, memory)
```

OUTPUT:
```python
0.0
```

# Solution ✅

````python
def evaluate_neuron(instructions, memory):
    """
    Evaluates a neuron based on the given instructions and memory.

    Memory is a dictionary.
    Instructions will form a stack.
    """
    stack = []

    # Instructions include actions, such as PUSH and OP,
    #   and targets, which constitute the keys in the memory.
    for action, target in instructions:
        if action == "PUSH":
            # Look up value in dictionary and push to stack
            value = memory[target]
            stack.append(value)

        elif action == "OP":
            if target == "*":
                val2 = stack.pop()
                val1 = stack.pop()
                stack.append(val1 * val2)

            elif target == "+":
                val2 = stack.pop()
                val1 = stack.pop()
                stack.append(val1 + val2)

            elif target == "RELU":
                val = stack.pop()
                stack.append(max(0.0, val))

    return stack.pop()
````

## Example run:

````python
evaluate_neuron(instructions, memory)
````
