## Solution

````python
"""
Below function is an implementation of the insertion sort algorithm in Python.
Resource:
https://realpython.com/sorting-algorithms-python/#implementing-insertion-sort-in-python

The function `sort_playlist` takes a `playlist` as input and sorts it in ascending order
using the insertion sort technique.
It also prints the sorted playlist to the console to "record each shift".
"""


def sort_playlist(playlist):
    print("Start:", playlist)
    # Loop from the second element of the array until the last element
    for i in range(1, len(playlist)):
        key_item = playlist[i]

        j = i - 1

        while j >= 0 and playlist[j] > key_item:
            # Shift the value one position to the left
            # and reposition j to point to the next element
            # (from right to left)
            playlist[j + 1] = playlist[j]
            j -= 1

        # When you finish shifting the elements, you can position
        # `key_item` in its correct location
        playlist[j + 1] = key_item
        print("Pass", i, ":", playlist)

    return playlist
````

Time complexity: $O(n^2)$

Space complexity: $O(1)$

````python
playlist = ["Song 2", "Superstition", "Sounds Like a Melody"]
sort_playlist(playlist)
````

## 2) Sort Detective 🕵️

Three students sorted the same list using three different sorting algorithms.

```text
Original list: [5, 2, 4, 1, 3]
The algorithms: Bubble sort, Selection sort, Insertion sort
```

Below are the traces after each main pass. 

```text
Trace A
Start:  [5, 2, 4, 1, 3]
Pass 1: [2, 4, 1, 3, 5]
Pass 2: [2, 1, 3, 4, 5]
Pass 3: [1, 2, 3, 4, 5]
Pass 4: [1, 2, 3, 4, 5]

Trace B
Start:  [5, 2, 4, 1, 3]
Pass 1: [1, 2, 4, 5, 3]
Pass 2: [1, 2, 4, 5, 3]
Pass 3: [1, 2, 3, 5, 4]
Pass 4: [1, 2, 3, 4, 5]

Trace C
Start:  [5, 2, 4, 1, 3]
Pass 1: [2, 5, 4, 1, 3]
Pass 2: [2, 4, 5, 1, 3]
Pass 3: [1, 2, 4, 5, 3]
Pass 4: [1, 2, 3, 4, 5]
```

Tasks
1. Match each trace to the correct sorting algorithm.
2. For each trace, write one sentence explaining your decision.
3. Which algorithm seems to benefit most from nearly sorted input?

### Solution

Trace A is __Bubble sort__.

In each pass, larger values move towards the end of the list. After the first pass, `5` has reached the last position.

Trace B is __Selection sort__.

In each pass, the smallest remaining value is selected and placed into its final position. After the first pass, `1` is already fixed at the beginning.

Trace C is __Insertion sort__.

The sorted part grows from left to right. Each new value is inserted into the correct position among the values already seen.

For all the mentioned sorting algorithms in this task:

Time complexity: $O(n^2)$

Space complexity: $O(1)$

__Task 3__: insertion sort benefits most from nearly sorted input because it only needs a few shifts when most elements are already close to their correct position.
